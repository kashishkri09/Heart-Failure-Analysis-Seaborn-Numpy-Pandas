# **Analysing the numeric column by using distribution plot

```python
import pandas as pd
import numpy as np
import seaborn as sns
df = pd.read_csv(r"C:\Users\Rishi kumar\Downloads\heart.csv")
df.head()
```

```python
#distribution plot
sns.distplot(df['Age'], kde = True) #kde plot line
```

```python
sns.distplot(df['Age'], kde = False, hist = True, rug = True)
```

```python
sns.distplot(df['Age'], kde = True, hist = True, rug = True)
```

```python
sns.distplot(df['Age'], bins = 5)
```

```python
sns.distplot(df['RestingBP'], kde = True)
```

```python
sns.distplot(df['Cholesterol'], kde = True, color='blue')
```

```python
sns.distplot(df['MaxHR'], kde = True, color='blue')
```

Use the Seaborn library to create a distplot of the "Age" column which will show the age distribution of individuals in the dataset.

Customize the plot to include a histogram and density plot and experiment with different bins to refine the visualization.
