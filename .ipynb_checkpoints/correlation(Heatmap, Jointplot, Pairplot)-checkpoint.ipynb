```python
#map of correlation using heatmap
sns.heatmap(df.corr(numeric_only = True))
```

Now we explore how to find the correlation between different numerical columns in the dataset using a heatmap. Correlation helps us understand whether an increase in one variable leads to an increase or decrease in another.

The values in the matrix are the Pearson correlation coefficients between the different variables, with 1 indicating a strong positive correlation, -1 indicating a strong negative correlation and 0 indicating no correlation. The colors in the heatmap represent the strength of the correlation, with darker colors indicating a stronger correlation.

Calculate the correlation matrix of the dataset using .corr().
Visualize the correlation using a heatmap to identify strong and weak relationships between features.
Analyze the correlation between age, cholesterol and heart disease, as well as other numerical variables like maximum heart rate.
The heatmap allows us to easily observe the relationships between variables and how they might impact heart disease predictions.

#  joint plot

```python
sns.jointplot(x='Age', y='MaxHR', data=df)
```

```python
sns.jointplot(x='Age', y='MaxHR', data=df, kind='hex')
```

```python
sns.jointplot(x='Age', y='MaxHR', data=df, kind='reg') #regression line(best fit line) how the data is flowing with age and maxhr
```

```python
sns.jointplot(x='HeartDisease', y='Oldpeak', data=df) #no corelated
```

```python
sns.jointplot(x='Cholesterol', y='Oldpeak', data=df) 
```

```python
sns.jointplot(x='Cholesterol', y='FastingBS', data=df, kind='reg') #NOT FIT LINE NO CORRELATION
```

```python
sns.jointplot(x='HeartDisease', y='MaxHR', data=df, kind='reg') #taking 2 column
```

```python
Now we dive deeper into analyzing the correlation between columns using a Joint Plot. This method helps visualize the relationship between two variables, especially to see if one variable increases or decreases as the other one changes. The Joint Plot gives both scatter and distribution plots, making it easy to identify patterns and correlations.

Use Seaborn's jointplot to plot the correlation between two columns, such as age and max heart rate (HR).
Plot a regression line to analyze the trend in the data.
Explore other correlations, like between heart disease and max HR, to visually understand how these variables are related.
Experiment with different plot types (scatter, hexbin) to get a better understanding of the relationships.
```
