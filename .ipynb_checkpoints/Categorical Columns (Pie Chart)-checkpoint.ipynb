# Categorical Columns (Pie Chart)/ use when we have less number of categories

```python
import pandas as pd
import numpy as np
import seaborn as sns
df = pd.read_csv(r"C:\Users\Rishi kumar\Downloads\heart.csv")
df.head()
```

```python
df.groupby('Sex').size().plot(kind = 'pie' , autopct= '%0.2f')
```

```python
df.groupby('ChestPainType').size().plot(kind = 'pie' , autopct= '%0.2f')
```

```python
df.groupby('RestingECG').size().plot(kind = 'pie' , autopct= '%0.2f')
```

```python
df.groupby('ST_Slope').size().plot(kind = 'pie' , autopct= '%0.2f')
```

```python
df.groupby('HeartDisease').size().plot(kind = 'pie' , autopct= '%0.2f')
```

We begin by using the groupby() function to categorize data based on the "Gender" column, then plot a pie chart to show the proportion of males and females in the dataset.
Similarly, we plot pie charts for the "Chest Pain Type" and "Heart Disease" columns to understand the distribution of different chest pain types and the presence or absence of heart disease.
The pie chart provides a clear visual understanding of the percentage distribution of these categories with easy interpretation of how many individuals fall into each category.
The analysis reveals that the dataset is skewed with a higher number of male individuals and a significant portion of people having heart disease.
